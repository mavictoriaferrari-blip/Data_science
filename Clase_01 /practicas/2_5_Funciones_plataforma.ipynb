funciones con retornos de valores

```python
def suma(a, b):
  return a + b
resultado = suma(5,3)
print (f"la suma de 5 y 3 es {resultado}")
```

Ejemplo

```python
def calcular_area_rectangulo (ancho,alto):
  return ancho * alto

ancho = 5
alto = 10
area = calcular_area_rectangulo (ancho, alto)
print(f"el area de un rectangulo de {ancho}x{alto} es {area}")
```
