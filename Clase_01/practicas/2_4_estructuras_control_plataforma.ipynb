estructuras de control

```python
x = 10
if x < 20:
  print ("x es menor que 20")
```

Estructura de control if-else

```python
x = 25
if x < 20:
  print ("x es menor que 20")
else:
  print ("x es mayor o igual a 20")
```

Esctructura de control for (bucle)

```python
for i in range(5):
  print(f"iteracion {i}")
```

estructura de control while

```python
i = 0
while i < 5:
  print(f"iteracion {i}")
  i += 1
```
